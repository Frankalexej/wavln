```python
import pandas as pd
import os
from paths import *
```

```python
csv_path = os.path.join(model_save_, "C_0Tf-0912014617-1", "guides", "ST-valid.csv")
df = pd.read_csv(csv_path)
```

```python


def shuffle_within_condition_and_speaker(df: pd.DataFrame, 
                                         speaker_col: str,
                                         condition_col: str,
                                         plosive_col: str,
                                         sibilant_cols: list,
                                         restrict_by_plosive: bool = True) -> pd.DataFrame:
    """
    Shuffle sibilant-related columns within the same condition and speaker, 
    optionally restricting to the same plosive.

    Args:
    df (pd.DataFrame): Input dataframe.
    speaker_col (str): Column indicating the speaker ID.
    condition_col (str): Column indicating the condition (e.g., 'sPV' or '#PV').
    plosive_col (str): Column indicating the plosive type.
    sibilant_cols (list): Columns related to the sibilant to be shuffled.
    restrict_by_plosive (bool): If True, restrict shuffling to same plosive.

    Returns:
    pd.DataFrame: DataFrame with shuffled sibilant columns.
    """
    
    df_copy = df.copy()
    
    # Group by speaker, condition, and optionally plosive
    group_cols = [speaker_col, condition_col]
    if restrict_by_plosive:
        group_cols.append(plosive_col)
    
    for _, group_idx in df_copy.groupby(group_cols).groups.items():
        group_df = df_copy.loc[group_idx]
        if len(group_df) <= 1:
            continue  # Skip groups with only one item
        
        # Shuffle sibilant columns
        shuffled_values = group_df[sibilant_cols].sample(frac=1, replace=False).values
        df_copy.loc[group_idx, sibilant_cols] = shuffled_values

    pre_length = df_copy['pre_endTime'] - df_copy['pre_startTime']
    df_copy['pre_endTime'] = df_copy['stop_startTime']
    df_copy['pre_startTime'] = df_copy['pre_endTime'] - pre_length
    return df_copy
```

```python
shuffled_df = shuffle_within_condition_and_speaker(
    df=df,
    speaker_col='speaker',
    condition_col='phi_type',
    plosive_col='stop',
    sibilant_cols=['pre', 'pre_path', 'pre_startTime', 'pre_endTime'],
    restrict_by_plosive=True  # Set False to shuffle across plosives
)
```
